 Y-Data-Profiling is a Python library designed to simplify Exploratory Data Analysis (EDA). It generates detailed reports for datasets, providing insights into data quality, statistics, distributions, correlations, missing values, and more—all with just a few lines of code.

```python
import pandas as pd
from ydata_profiling import ProfileReport

# Sample DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Tokyo']
})

# Generate profiling report
profile = ProfileReport(df, title="Profiling Report")
profile.to_file("report.html")
```
