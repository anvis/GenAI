To standardize data so that it has a mean of 0 and a variance of 1, you typically apply feature scaling using the Z-score normalization formula. Here's the step-by-step method.

You should standardize when:
The model relies on feature scaling (e.g., SVM, Logistic Regression, k-NN, Neural Networks).
Features have very different ranges (to prevent certain features from dominating).

```python
import numpy as np

# Example dataset
data = np.array([10, 20, 30, 40, 50])

# Standardize the data
mean = np.mean(data)        # Calculate mean
std_dev = np.std(data)      # Calculate standard deviation

standardized_data = (data - mean) / std_dev  # Apply Z-score formula

print("mean", mean)
print("Standard Deviation:", std_dev)
print("Standardized Data:", standardized_data)
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Sample dataset (features)
data = pd.DataFrame({
    'Feature1': [10, 20, 30, 40, 50],
    'Feature2': [15, 25, 35, 45, 55]
})
print(data)
# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the data
standardized_data = scaler.fit_transform(data)

# Convert back to a DataFrame for readability
standardized_df = pd.DataFrame(standardized_data, columns=data.columns)

print("Standardized Data:")
print(standardized_df)
```
